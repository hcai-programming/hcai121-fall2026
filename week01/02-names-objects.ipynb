# Names and Objects

So far we have been writing things like `x = 10` and saying "the variable `x` stores the value 10." That way of talking is fine for everyday use, but it is not quite how Python works.


- An **object** is a *bucket*. The bucket's contents are a value: the number ten, the text `"Ann"`, etc.
- A **variable** is a *label* stuck on the outside of a bucket. A bucket can wear many labels, and a label can be peeled off and stuck onto a different bucket.
- Crucially,  labels are *not* buckets. Variables don't contain values,  they point at objects that do.


## 1. `id()`

Every object (bucket) gets a **serial number** when it is created. `id(x)` reports the serial number of the bucket that the label `x` is currently stuck on.

There are two guarantees:

1. A bucket keeps the same serial number for its whole life.
2. No two buckets that exist *at the same time* share a serial number.

The actual number is meaningless. What matters is only whether two
serial numbers are **the same or different**.

````python
x = 10
print(id(x))
````

The label `x` is stuck on a bucket containing the number ten. `id(x)` is that bucket's serial number.

## 2. Two labels, one bucket

Will the two serial numbers below be the **same** or **different**?

````python
a = 10
b = 10
print(id(a))
print(id(b))
````

**Same!** There is only *one* bucket here, wearing *two* labels.

```
  a ──┐
      ├──►  [ bucket: 10 ]     ← one object, one serial number
  b ──┘
```

Assignment (`b = 10`) did not create a second copy of ten. Python noticed it already had a ten-bucket and stuck the label `b` on the same one.

*Warning: There is subtlety here, see the comments below.*

## 3.  `==` vs. `is`

Python has **two** comparison operators. They are easy to mix up, but they ask genuinely different questions:

| operator | question | in bucket language |
|---|---|---|
| `==` | same **value**? | do the buckets have the same *contents*? |
| `is` | same **object**? | is it literally the same *bucket*? |

`a is b` gives the same answer as `id(a) == id(b)`. 

````python
a = 10
b = 10
print(a == b)   # same contents?
print(a is b)   # same bucket?
````

## 4. Same value, different buckets 

For each line, do you predict it will display `True` or `False`?

````python
print(10.0 == 10)
print(10.0 is 10)
print(id(10.0))
print(id(10))
````

`10.0` is a **float** and `10` is an **int**. They can never be the same bucket, so the serial numbers always differ and `is` is always `False`.

But `==` compares *contents*, and Python's number types know how to compare across types.

This is the cleanest proof that **equality and identity are different questions**:

- `==` : equal contents — `True`
- `is` : same bucket — `False`, necessarily

*(You may also see Python print a warning about using `is` with a literal like `10`. That warning exists precisely because `is` on numbers is almost always a mistake — Section 5 shows why.)*

## 5. The fine print: don't rely on bucket-sharing

We have seen that `a = 10` and `b = 10` shared one bucket. It is tempting to conclude a rule: "equal values always share a bucket." **That rule is false.**

Run the next **two cells** and compare the serial numbers.

````python
big_a = 10000
print(id(big_a))
````

````python
big_b = 10000
print(id(big_b))
print(big_a == big_b)
print(big_a is big_b)
````

**Different buckets!** Equal contents (`==` is `True`), but two separate buckets (`is` is `False`).

What's going on? Python keeps a shelf of **pre-made buckets** for the small integers −5 through 256, because they get used constantly. Every `10` in your whole program gets its label stuck on the one
pre-made ten-bucket. But there is no pre-made bucket for `10000`, so Python manufactures one on demand, and in this case it manufactured two. 

(If you put both `10000` lines in the *same* cell, Python may notice it can reuse one bucket and you'll get `True`.  The point is that Python makes **no promise either way**.)

### The moral  

- Use `==` to compare **values**. This is almost always what you want.
- `is` has real uses, but never use it to compare numbers or text.

## 6. Reassignment moves the label, not the contents

What happens to the bucket when we assign a new value to `x`?

Will the serial number stay the same or change?

````python
x = 10
print(id(x))
x = 11
print(id(x))
print(id(10))
````

**It changes.** `x = 11` did **not** reach into the ten-bucket and repaint its contents.
It peeled the label `x` off the ten-bucket and stuck it on the eleven-bucket.

```
  before:   x ──►  [ bucket: 10 ]

  after:          [ bucket: 10 ]      ← untouched, just lost a label
             x ──►  [ bucket: 11 ]
```

Assignment in Python always moves labels,  it never edits buckets.


````python
votes = 100
print(id(votes))
votes = votes + 1        # count one more vote
print(votes)
print(id(votes))         # a different bucket!
````

### Strings work the same way

````python
candidate = "Ann"
print(id(candidate))
candidate = candidate + " Smith"
print(candidate)
print(id(candidate))     # a new bucket — the old "Ann" bucket was not edited
````

Even though it *looks* like we "added to" the string, Python built a brand-new string bucket
containing `"Ann Smith"` and moved the label. The original `"Ann"` bucket was never changed.


## 7. Immutable vs. mutable objects

Some objects are **immutable**: once created, their contents can never change. Everything we have met so far is immutable:

- `int`  (`10`)
- `float`  (`10.0`)
- `str`  (`"Ann"`)
- `bool`  (`True`, `False`)

With immutable objects, the *only* thing a program can ever do is move labels around. That is why every "change" in this notebook produced a new serial number: new value, new bucket, relocated label.


Soon we will meet **mutable** objects, in which contents *can* be edited in place (the same bucket, the same serial number, different contents over
time). 

## 8. Summary

- **Objects are buckets and variables are labels.** Labels are stuck on buckets, not stored in them.
- `id(x)` = the serial number of the bucket `x` is currently labeling. Serial numbers are unique among buckets that exist at the same time.
- `==` is true when the *contents are the same*
- `is` is true when the objects are the same.
- One bucket can wear many labels (`a = 10; b = 10`).
- Whether equal values share a bucket is something Python manages internally and not something you need to worry about. 
- Assignment **rebinds**: it moves a label to a different bucket. It never edits a bucket.
- Ints, floats, strings, and bools are **immutable**.  We will discuss **mutable** objects later in the course.
