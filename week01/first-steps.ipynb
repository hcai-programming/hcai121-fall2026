## 2. Variables

A **variable** is a name that refers to a value. We create one with `=`.

Read `price = 4` as a command: *"make the name `price` refer to the value 4."* It is **not** a claim that they are equal, the way it is in mathematics.


```python
price = 4
```

Notice that nothing printed. Assigning a variable is an *action*, not a question. To see the value, we ask for it:


```python
price
```

Now we can compute with the names:


```python
quantity = 3
total = price * quantity
total
```

## 3. `=` is not algebra

In a math class, `x = 5` and `x = 10` would be a contradiction. In Python, the second line simply re-points the name at a new value.


```python
x = 5
x = 10
x
```

Predict the value of `b` before running the following code.

## 4. Types of values

Every value in Python has a **type**. We can ask for it with `type()`.

| English name | Type name | Description | Example |
| :----------- | :-------- | :---------- | :------ |
| integer | `int` | whole numbers | `42` |
| floating point number | `float` | numbers with a decimal point | `3.14159` |
| string | `str` | text | `"Logic is fun!"` |
| boolean | `bool` | true or false | `True` |


```python
type(42)
```

```python
type(3.14159)
```

```python
type("Logic is fun!")
```

When we divide two whole numbers. What *type* is the answer?


```python
10 / 5
```

```python
type(10 / 5)
```

Dividing with `/` *always* gives a `float`, even when it divides evenly. If you want a whole number, use floor division `//`:


```python
25 // 2   # floor division: rounds down to a whole number
```

```python
25 % 2    # modulo: the remainder when 25 is divided by 2
```

## 5. Strings

A **string** is text, written inside single or double quotes (there is no difference between them):

- single quotes: `'Philosophy'`
- double quotes: `"Artificial Intelligence"`


```python
course = "HCAI 121"
course
```

What does `+` do with two strings?


```python
"Hello" + " " + "World"
```

So `+` *joins* strings. But watch what happens when the two sides are different types.


## 6. When types collide

Let's build the course name from two pieces:


```python
course_name = "HCAI"
course_num = 121
```

Will the following output HCAI 121? 


```python
course_name + " " + course_num
```

Read the error message: it is telling you that you can't join a string and a number. The fix is to *convert* the number to a string first, with `str()`:


```python
course_name + " " + str(course_num)   # convert the number to a string
```

Reading error messages is a real skill. An error is not a disaster; it is Python telling you exactly what went wrong.
