# Functions


A **function** is a named piece of work. You give it some inputs, it does something, and it may hand a result back. Just as a *variable* names a value, a *function* names a computation you can reuse.

As always, try to **predict** what each cell does before you run it.


## 1. The idea of a function

Think of a function as a little machine: something goes **in**, the machine does its work, and something may come **out**.

You have already been using functions! `print`, `type`, and `round` are all functions. When you write `round(2.5)`, you are *calling* the function `round` with the *input* `2.5`.

## 4. Writing your own functions

We can define our own functions with `def`:

```python
def greet(name):
    return "Hello, " + name + "!"
```

- `def` starts the definition
- `greet` is the function's **name**
- `name` is a **parameter** — an input the function expects
- the indented line is the **body**
- `return` hands a value back to whoever called the function


````python
def greet(name):
    return "Hello, " + name + "!"
````


Defining a function does **not** run it. The body runs only when we **call** the function — its name, with an argument in parentheses:


````python
greet("Ada")
````

````python
greet("Grace")
````

## 5. `return` vs. `print`

We saw that `print` shows but hands back `None`. This is the single most important idea about functions. Compare:


````python
def add_return(a, b):
    return a + b

def add_print(a, b):
    print(a + b)
````

Both seem to show the answer. But now let's try to **use** the result. **Predict each of the next two cells.**


````python
result = add_return(3, 4)
result * 2
````

````python
result = add_print(3, 4)
result * 2   # what is result here?
````

`add_print` displays `7` but hands back `None`, so `result` is `None` and `result * 2` fails. `add_return` hands the value back, so we can keep computing with it.

**`print` shows a value to a human. `return` gives a value back to the program.** A function that only prints cannot be used to build something larger.

## 7. Local and global variables

A variable created **inside** a function is **local**: it exists only while the function runs. **Predict:** will the last line work?


````python
def make_message(name):
    greeting = "Hi " + name   # greeting is local to this function
    return greeting

make_message("Sam")
````

````python
greeting   # error: greeting only exists inside make_message
````

A variable defined **outside** all functions is **global**, and a function is allowed to *read* it:


````python
tax_rate = 0.06        # a global variable

def price_with_tax(price):
    return price * (1 + tax_rate)   # reads the global tax_rate

price_with_tax(100)
````

This works, but relying on globals makes a function harder to understand and test, because its result depends on something outside it. **Good practice: pass in what you need as parameters** rather than reaching out to global variables.


## 8. What happens to the caller's variables?

When we pass a variable to a function, can the function change it? **Predict the value of `x` after this runs.**


````python
def add_one(n):
    n = n + 1     # change the parameter inside the function
    return n

x = 5
add_one(x)
````

````python
x   # did x change?
````

`x` is still `5`. Inside the function, `n` is a **separate name**; reassigning `n` does not reach back and change the caller's `x`. For the values we've used so far — numbers, strings, booleans — a function cannot change the caller's variables. (You may hear this called *pass by value*.)



## 9. Functions built from other functions

Functions can call other functions, so we can build big things from small, understandable pieces.


````python
def loud(text):
    return text.upper() + "!"

def loud_greeting(name):
    return loud(greet(name))   # uses both greet and loud
````

````python
loud_greeting("Ada")
````

Breaking a problem into small functions and combining them is the heart of programming — and exactly what your first project will ask you to do.


## 10. Named and default arguments

So far we have passed arguments **by position**: the first value fills the first parameter, and so on. There are two nicer ways to call a function.


### Named arguments

Recall our tax function:


````python
def total_cost(price, quantity):
    return price * quantity * 1.06
````

Reading `total_cost(10, 3)` cold, which number is the price and which is the quantity? You have to go check the definition. We can make the call explain itself by **naming** the arguments:


````python
total_cost(price=10, quantity=3)
````

**Predict:** with names, does the order still matter?


````python
total_cost(quantity=3, price=10)   # same result?
````

When we name the arguments, Python matches them by **name**, not by position, so the order no longer matters, and the call is easier to read.


### Default arguments

Right now the tax rate `1.06` is fixed. Suppose most orders use 6% tax, but some are tax-exempt. We can give a parameter a **default value**, used unless the caller supplies its own:


````python
def total_cost(price, quantity, tax_rate=0.06):
    return price * quantity * (1 + tax_rate)
````

**Predict each of these three calls.**


````python
total_cost(10, 3)                 # no tax_rate given -> uses the default 0.06
````

````python
total_cost(10, 3, 0.00)           # override: tax-exempt
````

````python
total_cost(10, 3, tax_rate=0.09)  # override with a name, for clarity
````

The `tax_rate=0.06` in the definition is a **fallback**: the caller can leave it out and get the default, or supply a different value. This "sensible default, override when needed" pattern is very common in Python.

One rule: parameters *with* a default must come **after** parameters without one. So `def f(a, b=1)` is fine, but `def f(a=1, b)` is an error.
