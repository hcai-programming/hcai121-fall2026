## 1. What is each part of a function *for*?


Here is the vocabulary, pinned to one example:


```python
def double(number):
    return number * 2
```

```python
double(8)
```

| Part | In the example | What it is |
| :--- | :--- | :--- |
| `def` | `def` | the keyword that *creates* a function |
| function name | `double` | the name we call it by |
| **parameter** | `number` | a *placeholder* name, used inside the body |
| body | `return number * 2` | the work the function does |
| **argument** | `8` | the *actual value* we pass in when we call it |
| return value (output) | `16` | the value handed back |

When you write `double(8)`, Python sets `number = 8` for that call, runs the body, and hands back `16`.  So, the **parameter** is the empty slot in the *definition* of the function, and the **argument** is what you drop into that slot when you *call* the function. 


## 2. What are the rules for naming a function?

> *"Are there other restrictions on naming functions, like there are with variables?"*

Yes, function names follow the **same rules as variable names**:

- use letters, digits, and underscores (`_`)
- must **not start with a digit**
- **no** dashes, spaces, or other special characters
- they are case-sensitive (`Double` and `double` are different)
- they cannot be a **reserved word** (a word Python already uses, like `def`, `return`, `if`, `for`, `class`)

So these are all errors: `def 2square():`, `def my-func():`, `def def():`.

You can see the full list of reserved words:


```python
import keyword
print(keyword.kwlist)
```

## 3. `print` vs. `return`

> *"I understand print is another word for submit or running the program, so I'm confused how you can return a function."*

There are two things that need to be cleared up:

- `print` is **not** "running the program." `print` is just a function that **shows text on the screen**. Running the program is what happens when Python executes your cells,  `print` is one small thing that *can happen* while it runs.
- `return` does not "return a function." It returns a **value** *from* a function.   That is,  it hands a value back to whoever called it.

The clean distinction:

- **`print`** shows a value to a *human*. It puts text on the screen and then is done.
- **`return`** gives a value back to the *program*, so you can store it in a variable and keep using it.


Here is a function that **returns**:


```python
def square_return(n):
    return n * n

y = square_return(5)
y * 2          # we can keep computing with y, because we got the value back
```

Here is one that only **prints**:


```python
def square_print(n):
    print(n * n)

y = square_print(5)   # this shows 25 on the screen...
y                     # ...but what is y?
```

`square_print(5)` **shows** `25`, but it never *hands a value back*, so `y` is `None`. The number `25` was displayed and then gone.  It was never given to `y`.


## 4. Why is the result `None`?

Recall the following example: 

```python
def square(n):
    print(n * n)

y = square(5)
print(y)
```

> *"I thought `y` had already been assigned 25, but when the code prints `y`, the result is `None`."*

You see `25` (from the `print` *inside* the function) and then `None` (the value of `y`). Because `square` only *prints* and has no `return`, calling it hands back `None`. So,  `y` is `None`, not `25`.

If not return value is defined, functions in Python return `None`.  In the falling example `pass` is a keyword that tells Python that there is no body to execute. 

```python
def f(x):
    pass

y = f(5)
print(y)
```

The fix to the above `square` function is to **return** the value instead of printing it:


```python
def square(n):
    return n * n

y = square(5)
print(y)        # now y really is 25
```

## 5. Local variables

> *"I get a little confused on the local variables part."*

A variable created **inside** a function is **local**: it exists only while that function is running, and it cannot be seen from outside. Here are two functions (this is the `print_twice` example):


```python
def print_twice(text):
    print(text)
    print(text)

def join_and_print(part1, part2):
    combined = part1 + part2   # combined is LOCAL to this function
    print_twice(combined)
```

```python
join_and_print("HCAI", "121")
```

`combined` did its job inside `join_and_print` and then disappeared. 

```python
combined   # error: combined only exists inside join_and_print
```

Note: This behavior is a *good* thing: each function gets its own private workspace, so the names you use inside one function can't accidentally clash with names elsewhere. 

## 6. Are you allowed to use multiple parameters with defaults, and does the order of those matter?

 
Yes, you can have as many defaults as you like. The one rule is: **every parameter with a default must come after every parameter without one.**


```python
def order(item, size="medium", extra="none"):
    return item + " (" + size + ", extra: " + extra + ")"
```

Now you can call it several ways. **Predict each one.**


```python
order("coffee")                         # both defaults used
```

```python
order("coffee", "large")                # override size only
```

```python
order("coffee", "large", "sugar")       # override both
```

About order: when you pass arguments **by position**, they fill the parameters left to right, so `order("coffee", "large")` sets `size`, not `extra`. To skip an earlier default and set a later one, use the parameter's **name**:


```python
order("coffee", extra="sugar")          # keep size default, set extra
```

So: multiple defaults are fine; among themselves you can list them in whatever order makes sense; and naming your arguments (`extra="sugar"`) lets you set just the ones you want, in any order, and leave the rest at their defaults.
