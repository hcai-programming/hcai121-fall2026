# Functions and Interfaces


Once we can write functions, the next question is: how do we design *good* ones, and how do we build them up from each other? This notebook follows the ideas in *Think Python*, Chapter 4 using small number-and-string examples.


## 1. Encapsulation: wrapping code in a function

Suppose we compute the total cost of an order with 6% tax:


```python
price = 10
quantity = 3
price * quantity * 1.06
```

Wrapping this up in a function is called **encapsulation**. It gives the computation a name (which documents what it does) and lets us reuse it without copying and pasting.


```python
def total_cost():
    return 10 * 3 * 1.06
```

```python
total_cost()
```

## 2. Generalization: adding parameters

The function above always uses the same numbers. **Generalization** means replacing specific values with parameters, so the function becomes more flexible:


```python
def total_cost(price, quantity):
    return price * quantity * 1.06
```

```python
total_cost(10, 3)
```

```python
total_cost(4, 5)
```

With a parameter, the same function handles any order. This is the same move in Chapter 4 where a fixed-size `square()` is turned into `square(length)`.


## 3. Interface vs. implementation

A function has two parts:

- Its **interface** is *how you use it*: its name, its parameters, and what it does.
- Its **implementation**: the code inside.

Here are two versions of a function that adds tax to a price. They compute the answer differently, but they are called the same way and give the same result.


```python
def with_tax(price):
    return price * 1.06
```

```python
def with_tax(price):
    tax = price * 0.06
    return price + tax
```

```python
with_tax(100)
```

Same **interface** (`with_tax(price)`, returns the price plus 6% tax); different **implementation** (the code inside). Because the interface didn't change, anything that *calls* `with_tax` keeps working even though we rewrote the inside.

This is why we can improve a function's guts without breaking the rest of the program, an idea we'll rely on all semester.


## 4. Refactoring

**Refactoring** means reorganizing working code to make it better, without changing what it does. Often we notice a smaller piece that could be its own function, and pull it out.

For example, we might pull the tax calculation into its own function and have `total_cost` use it:


```python
def with_tax(price):
    return price * 1.06

def total_cost(price, quantity):
    return with_tax(price) * quantity
```

```python
total_cost(10, 3)
```

The behavior is the same as before, but now the tax logic lives in one place. Building functions out of other functions like this is the heart of good program design.


## 5. Docstrings: documenting the interface

A **docstring** is a string at the top of a function that explains its interface: what it does and what to pass in. By convention it is written with triple quotes:


```python
def total_cost(price, quantity):
    """Return the total cost of an order, including 6% tax.

    price: the price of one item
    quantity: how many items are ordered
    """
    return with_tax(price) * quantity
```

The docstring describes *what the function does and what to give it* — the interface — not the details of *how* it works. Python even shows it to you with `help`:


```python
help(total_cost)
```

A good sign of a well-designed function: its interface is easy to explain. If you struggle to write the docstring, the function might be trying to do too much.
