# Booleans, Logic, and Conditionals

So far our programs run every line, top to bottom. Now we give them the ability to **make decisions**: to do one thing in some cases and something else in others. The foundation for this is the boolean values `True` and `False`.


## 1. Boolean values

A **boolean** is one of just two values: `True` or `False`. Its type is `bool`.


```python
True
```

```python
type(True)
```

Note the capital `T` and `F`, and no quotes: `True` is the boolean, while `"True"` would just be a string.


## 2. Comparisons ask a yes/no question

**Comparison operators** compare two values and produce a boolean:

| Operator | Meaning |
| :------: | :------ |
| `==` | equal to |
| `!=` | not equal to |
| `<` | less than |
| `>` | greater than |
| `<=` | less than or equal to |
| `>=` | greater than or equal to |


```python
5 == 5
```

```python
5 != 6
```

```python
3 < 7
```

Are these two strings equal?


```python
"cat" == "Cat"
```

They are **not** equal:  comparisons on strings are case-sensitive, so `"cat"` and `"Cat"` are different.


### A crucial distinction: `=` vs. `==`

These look similar but do completely different things:

- `=` **assigns**: `x = 5` means *make `x` refer to 5*.
- `==` **compares**: `x == 5` means *is `x` equal to 5?*, and gives back `True` or `False`.

One is a command; the other is a question.


```python
x = 5     # assignment: set x to 5
x == 5    # comparison: is x equal to 5?
```

Mixing them up is one of the most common beginner mistakes. (Writing `if x = 5:` is actually an error in Python.)


## 3. Combining conditions: `and`, `or`, `not`

We often need to check more than one thing. The **logical operators** combine booleans:

- `A and B` is `True` only when **both** are `True`.
- `A or B` is `True` when **at least one** is `True`.
- `not A` **flips** it: `not True` is `False`.


```python
A = True
B = True
print("When A is True and B is True,  A and B is ", A and B)
A = True
B = False
print("When A is True and B is False,  A and B is ", A and B)
A = False
B = True
print("When A is False and B is True,  A and B is ", A and B)
A = False
B = False
print("When A is False and B is False,  A and B is ", A and B)
```

```python
A = True
B = True
print("When A is True and B is True,  A or B is ", A or B)
A = True
B = False
print("When A is True and B is False,  A or B is ", A or B)
A = False
B = True
print("When A is False and B is True,  A or B is ", A or B)
A = False
B = False
print("When A is False and B is False,  A or B is ", A or B)
```

Here is a real example, checking whether an age is in the teenage range:


```python
age = 15
age >= 13 and age <= 19
```

Python even lets you chain comparisons the way math does:


```python
13 <= age <= 19
```

### A common mistake with `or`

Suppose we want to check whether `day` is Saturday or Sunday. It is tempting to write it like English. **Predict:** what does this print when `day` is `"mon"`?


```python
day = "mon"
day == "sat" or "sun"
```

Surprise: it gives `"sun"`, which counts as `True`! Python reads this as `(day == "sat") or ("sun")` — and a non-empty string like `"sun"` is treated as `True`. So this condition is **always** true, no matter what `day` is.

You must compare `day` to each value **separately**:


```python
day == "sat" or day == "sun"
```

## 4. Making decisions with `if`

An `if` statement runs a block of code **only when** a condition is `True`. The block is marked by **indentation** (the spaces at the start of the line):


```python
temperature = 90

if temperature > 80:
    print("It's hot!")
```

Add an `else` for what to do otherwise:


```python
temperature = 60

if temperature > 80:
    print("It's hot!")
else:
    print("It's not hot.")
```

For several cases, use `elif` (short for "else if"). Python checks the conditions in order and runs the **first** one that is true:


```python
def classify(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    else:
        return "F"
```

```python
classify(85)
```

```python
classify(95)
```

```python
classify(50)
```

## 5. A surprise: separate `if`s vs. `elif`

Here is the same grading idea written with **separate `if` statements** instead of `elif`. **Predict what it prints for a score of 85.**


```python
score = 85

if score >= 90:
    grade = "A"
if score >= 80:
    grade = "B"
if score >= 70:
    grade = "C"

print(grade)
```

It prints `"C"`, not `"B"`! With separate `if`s, Python checks **every** one. For score 85, both `>= 80` and `>= 70` are true, so `grade` is set to `"B"` and then overwritten with `"C"`, the last true one wins.

This is exactly why we use `elif`: it stops at the **first** true condition. The `classify` function above gets it right for the same reason.
