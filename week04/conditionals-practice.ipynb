## 3. Combining conditions


**Predict:**


````python
temp = 75
print(70 <= temp <= 80)
````

**Predict:** what prints? 


````python
color = "red"
if color == "blue" or "green":
    print("cool color")
else:
    print("warm color")
````

**Why:** Python reads this as `(color == "blue") or ("green")`. The first part is `False`, but `"green"` is a non-empty string, which counts as `True`, so the whole condition is **always** true, no matter the color. The fix is to compare fully on both sides: `color == "blue" or color == "green"`.


## 4. if / elif / else


**Predict** both calls:


````python
def classify(n):
    if n > 0:
        return "positive"
    elif n < 0:
        return "negative"
    else:
        return "zero"

print(classify(-4))
print(classify(0))
````

**Predict:** what label prints? 


````python
temp = 85
if temp >= 80:
    label = "hot"
if temp >= 60:
    label = "warm"
if temp >= 40:
    label = "cool"
print(label)
````

**Why:** these are **separate** `if` statements, so Python checks *all* of them. For 85, all three conditions are true, so `label` becomes `"hot"`, then is overwritten with `"warm"`, then `"cool"`, the last true one wins. Using `elif` instead would stop at the first match and correctly give `"hot"`.


You can put an `if` inside another `if`. **Predict** both calls:


````python
def entry(age, has_ticket):
    if has_ticket:
        if age < 12:
            return "child, go in"
        else:
            return "adult, go in"
    else:
        return "no ticket"

print(entry(10, True))
print(entry(30, False))
````

**Predict:** 


````python
temp = 30
if temp >= 80:
    weather = "hot"
elif temp >= 60:
    weather = "warm"
print(weather)
````

**Why:** no branch ran (temp is 30), so `weather` was never created, and printing it is a `NameError`. Adding an `else` that always sets `weather` avoids this.

## 6. A shortcut: conditional expressions

Python has a compact way to **choose between two values** based on a condition:

```python
x = value_if_true if condition else value_if_false
```

Read it left to right: *x gets `value_if_true` if the condition holds, otherwise `value_if_false`.*


**Predict:**


````python
age = 20
status = "adult" if age >= 18 else "minor"
print(status)
````

This does exactly the same thing as the longer form below:

```python
if age >= 18:
    status = "adult"
else:
    status = "minor"
```

Because it's an **expression** (it produces a value), you can even use it directly inside a `print` or a `return`. **Predict** these:
