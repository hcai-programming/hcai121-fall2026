# Loops

A quick overview of `for` loops, `while` loops, and looping over strings. Run the setup cell first. We will work through these together in class.

```python
import random

def random_move():
    n = random.randint(1, 3)
    if n == 1:
        return "rock"
    elif n == 2:
        return "paper"
    else:
        return "scissors"

def winner(player, computer):
    if player == computer:
        return "tie"
    elif (player == "rock"     and computer == "scissors") or \
         (player == "paper"    and computer == "rock")     or \
         (player == "scissors" and computer == "paper"):
        return "player"
    else:
        return "computer"
```

## For loops

**Q1.** What does this print?

```python
for i in range(4):
    print(i)
```

**Q2.** What is the value of `total` at the end?

```python
total = 0
for i in range(5):
    total = total + 2
print(total)
```

**Q3.** What does this print, and why?

```python
wins = 0
for i in range(10):
    if winner("rock", "paper") == "player":
        wins = wins + 1
print(wins)
```

## While loops

**Q4.** What does this print?

```python
i = 0
while i < 3:
    print(i)
    i = i + 1
```

**Q5.** What does this print?

```python
count = 0
n = 10
while n > 1:
    n = n - 3
    count = count + 1
print(n, count)
```

## Looping over strings

**Q8.** A string is a sequence of characters, so `in` can check whether a character is inside it. What does each of these return?

```python
print("a" in "hcai")
print("z" in "hcai")
```

**Q9.** A `for` loop can go through a string one character at a time. What does this print?

```python
for letter in "hcai":
    print(letter)
```

**Q10.** This loop searches the string for the letter `"a"`. What does it print? (This is what `"a" in "hcai"` does behind the scenes.)

```python
found = False
for letter in "hcai":
    if letter == "a":
        found = True
print(found)
```
