# Rock, Paper, Scissors: Extensions

We built the game with **functions** and **conditionals**. Now we extend it:

1. Read the player's move with `input`.
2. Check the move with `validate_input`.
3. Play several rounds with a `for` loop.
4. Keep asking until the move is valid with a `while` loop.
5. Keep win and loss records.
6. Write a function to test a strategy over many rounds.

## Setup

Run this once. It defines the computer's move, the `winner` rule, and `play` for one round. Paste in your team's `winner()` if you would rather use yours.

```python
import random

def random_move():
    n = random.randint(1, 3)
    if n == 1:
        return "rock"
    elif n == 2:
        return "paper"
    else:
        return "scissors"

def winner(player, computer):
    if player == computer:
        return "tie"
    elif (player == "rock"     and computer == "scissors") or \
         (player == "paper"    and computer == "rock")     or \
         (player == "scissors" and computer == "paper"):
        return "player"
    else:
        return "computer"

def play(player):
    computer = random_move()
    print("You chose:     ", player)
    print("Computer chose:", computer)

    result = winner(player, computer)
    if result == "tie":
        print("It's a tie!")
    elif result == "player":
        print("You win!")
    else:
        print("Computer wins!")
```

## 1. Ask the player for a move

`input` reads what the user types and returns it as a string. Run this and type `rock`, `paper`, or `scissors`.

```python
player = input("Enter rock, paper, or scissors: ")
play(player)
```

## 2. Check the move with `validate_input`

What if someone types `"Rock"` or `"banana"`? Right now `play` runs anyway and the winner rule quietly says the computer wins. We want to catch a bad move first.

Write a function that returns `True` for a valid move and `False` otherwise.  

```python
def validate_input(player):
    if player == "rock" or player == "paper" or player == "scissors":
        return True
    else:
        print("Invalid input. Please enter rock, paper, or scissors.")
        return False

print(validate_input("rock"))
print(validate_input("test"))
```

Now check the move before playing:

```python
player = input("Enter rock, paper, or scissors: ")
if validate_input(player):
    play(player)
```

## 3. Play several rounds with a `for` loop

To play more than once, use a **loop**. `range(3)` gives the numbers `0, 1, 2`, and the indented body runs once for each. Predict the output, then run.

```python
for i in range(3):
    print("round", i)
```

Put `play` inside the loop to play three rounds in a row:

```python
for i in range(3):
    play("rock")
    print()
```

## 4. Keep asking until the move is valid with a `while` loop

A `for` loop runs a fixed number of times. But we do not know how many times a user will type a bad move. A `while` loop keeps going as long as its condition is `True`.

We ask once, then keep asking `while` the move is `not` valid. When a valid move finally comes in, the condition is `False` and the loop stops.

```python
player = input("Enter rock, paper, or scissors: ")
while not validate_input(player):
    player = input("Enter rock, paper, or scissors: ")

play(player)
```

## 5. Play until you quit with a `while` loop

Another use of `while`: keep playing rounds until the player wants to stop. Start with `answer = "yes"` and loop `while answer == "yes"`. The inner `while` is the input check from Section 4.

Run it, and type `no` when you want to stop.

```python
answer = "yes"
while answer == "yes":
    player = input("Enter rock, paper, or scissors: ")
    while not validate_input(player):
        player = input("Enter rock, paper, or scissors: ")
    play(player)
    answer = input("Play again? (yes/no): ")
```

## 6. First to 3 wins

A match keeps going until someone has won 3 rounds. That is a `while` loop with a condition on the score: keep playing `while` both scores are below 3.

**Your turn.** Fill in the score update. Play one round each time through the loop, and stop when the player or the computer reaches 3 wins.

```python
player_wins = 0
computer_wins = 0

while player_wins < 3 and computer_wins < 3:
    player = input("Enter rock, paper, or scissors: ")
    while not validate_input(player):
        player = input("Enter rock, paper, or scissors: ")

    computer = random_move()
    result = winner(player, computer)
    print("Computer chose:", computer)

    # your turn: add 1 to player_wins if result is "player",
    # or 1 to computer_wins if result is "computer" (a tie adds nothing).
    # then print the running score.

print("Game over. You:", player_wins, "Computer:", computer_wins)
```

## 7. Keep win and loss records

Now play several rounds and count the results. Start three counters at `0`, and add `1` to one of them each round.

```python
wins = 0
losses = 0
ties = 0

for i in range(20):
    computer = random_move()
    result = winner("rock", computer)
    if result == "player":
        wins = wins + 1
    elif result == "computer":
        losses = losses + 1
    else:
        ties = ties + 1

print("wins:", wins)
print("losses:", losses)
print("ties:", ties)
```
