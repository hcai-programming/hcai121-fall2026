## A list of lists

A list can hold other lists. That gives you a grid (a table of rows). Reach an item with two indexes: `grid[row][col]`.

```python
grid = [[1, 2, 3],
        [4, 5, 6]]

print(grid[0])       # the first row
print(grid[1][2])    # row 1, column 2
print(len(grid))     # number of rows
print(len(grid[0]))  # number of columns in a row
```

## Looping over a grid

Loop over the rows, and loop again inside each row to reach every item.

```python
grid = [[1, 2, 3],
        [4, 5, 6]]

for row in grid:
    print(row)

total = 0
for row in grid:
    for x in row:
        total = total + x
print(total)
```

## Building a grid

Build the rows with a comprehension. Be careful: `[[0] * 3] * 2` looks right but makes two names for the *same* inner list, so changing one row changes the other.

```python
good = [[0] * 3 for _ in range(2)]
good[0][0] = 9
print(good)     # only the first row changed

bad = [[0] * 3] * 2
bad[0][0] = 9
print(bad)      # both rows changed!
```

## Copying a list of lists

A slice `[:]` (or `list(...)`) copies the outer list but still shares the inner lists. For a fully independent copy, use `copy.deepcopy`.

```python
import copy

a = [[1, 2], [3, 4]]
shallow = a[:]        # outer copied, inner lists shared
shallow[0][0] = 99
print(a)              # a changed too!

a = [[1, 2], [3, 4]]
deep = copy.deepcopy(a)
deep[0][0] = 99
print(a)              # a is safe
```

## A list of tuples

Each item can be a tuple, which is handy for records like `(name, score)`. Unpack the tuple right in the loop.

```python
scores = [("alice", 90), ("bob", 85), ("carol", 95)]

for name, score in scores:
    print(name, score)

total = 0
for name, score in scores:
    total = total + score
print(total)
```

## Sorting records

`sorted` on a list of tuples orders by the first item. To sort by the score instead, give it a `key`. The `key=...` line is a preview; we cover `key` and `lambda` later.

```python
scores = [("alice", 90), ("bob", 85), ("carol", 95)]
print(sorted(scores))                            # by name
print(sorted(scores, key=lambda pair: pair[1]))  # by score
```
