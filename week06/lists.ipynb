## A list is a sequence

A list is a sequence of values written with square brackets. Unlike a string, a list can hold values of any type. `len` gives the number of items.

````python
x = "hcai"

len(x)
````

````python
numbers = [42, 123]
wines = ["Merlot", "Malbec", "Riesling"]
empty = []
lst = [1, 2, "hello", 3.14, True]

print(wines)
print(len(wines))
print(len(empty))
print(len(lst))
````

## Getting items by index

Each item has an index, starting at `0`. A negative index counts from the end.

````python
wines = ["Merlot", "Malbec", "Riesling"]
print(wines[0])
print(wines[2])
print(wines[-1])
print(len(wines))
print(wines[len(wines) - 1])
````

## Lists are mutable

Unlike a string, you can change an item by assigning to its index. `in` checks whether a value is in the list.

````python
numbers = [42, 123]
numbers[1] = 17
print(numbers)

print("Malbec" in wines)
print("Chardonnay" in wines)
````

````python
x = "hello"
y = x
y = "goodbye"
print("x is ", x)
print("y is ", y)
````

````python
numbers = [42, 123]
numbers2 = numbers
print(numbers is numbers2)
numbers2[0] = 99
print("numbers:", numbers)
print(numbers is numbers2)

numbers[1] = 17

print("numbers:", numbers)
print("numbers2:", numbers2)
````

## Slices

A slice gives you part of a list, the same way it does with a string.

````python
letters = ["a", "b", "c", "d"]
print(letters[1:3])
print(letters[:2])
print(letters[2:])

print("hcai"[0:2])
print("hcai"[2:4])
print("hcai"[-2:])
````

## Adding lists together

`+` joins two lists into a new one. `*` repeats a list.

````python
print([1, 2] + [3, 4])
print(["a"] * 3)
print([0] * 5)
````

## Adding and removing

`append` adds one item to the end. `pop` removes and returns an item by index. `remove` deletes an item by value.

````python
letters = ["a", "b", "c"]
letters.append("d")
print(letters)

first = letters.pop(0)
print(first, letters)

letters.remove("c")
print(letters)
````

Careful: `append` and `remove` change the list and return `None`. Do not write `letters = letters.append("e")`, or you will lose your list.

````python
letters = ["a", "b", "c"]
result = letters.append("d")
print(result)    # None
print(letters)   # but the list did change
````

## More ways to change a list

`insert` adds an item at a position. `extend` adds several items at once. `sort` puts the list in order, in place. `del` removes an item by index.

````python
letters = ["a", "b", "c"]
letters.insert(1, "x")
print(letters)

letters.extend(["y", "z"])
print(letters)

del letters[0]
print(letters)

nums = [3, 1, 2]
nums.sort()
print(nums)
````

## Lists and strings

`list` turns a string into a list of characters. `split` breaks a string into a list of words. `join` glues a list of strings back together.

````python
print(list("spam"))
print("pining for the fjords".split())
print("-".join(["hcai", "121", "fall", "2026"]))
````

## Looping over a list

A `for` loop goes through a list one item at a time, just like it does with a string.

````python
for wine in wines:
    print(wine)
````

## Building a list with a loop

Start with an empty list and `append` one item at a time. `range` gives you a sequence of numbers to loop over, and `list` shows what it contains.

````python
squares = []
for i in range(5):
    squares.append(i * i)
print(squares)

print(list(range(5)))
print(list(range(2, 10, 2)))
````

## List comprehensions

A list comprehension builds a list in one line. It does the same thing as the loop above.

````python
squares = [i * i for i in range(5)]
print(squares)

print([2 * i for i in range(10)])
print([w.upper() for w in wines])
````

You can add an `if` to keep only some of the items.

````python
print([n for n in range(10) if n % 2 == 0])
print([w for w in wines if len(w) == 6])
````

## sorted

`sorted` returns a new sorted list and leaves the original alone. Compare `list.sort` above, which changes the list in place and returns `None`.

````python
scramble = ["c", "a", "b"]
print(sorted(scramble))
print(scramble)
````

### `sort()` vs `sorted()`

`sorted(xs)` makes a **new** sorted list and leaves `xs` alone. `xs.sort()` sorts `xs` **in place** and returns `None`. Predict each line before you run it.

````python
nums = [3, 1, 2]

new = sorted(nums)      # a brand new sorted list
print(new)              # [1, 2, 3]
print(nums)             # [3, 1, 2]  -- nums is unchanged

answer = nums.sort()    # sorts nums in place, returns None
print(answer)           # None  -- so never write  nums = nums.sort()
print(nums)             # [1, 2, 3]  -- nums itself changed
````

## Some useful functions

`sum`, `min`, `max`, and `len` all work on a list of numbers.

````python
nums = [4, 8, 15, 16, 23, 42]
print(sum(nums))
print(min(nums))
print(max(nums))
print(len(nums))
````

## Two names for the same list

Assigning a list does not copy it. Both names point at the same list, so a change through one shows up in the other. `is` asks whether two names point at the same list; `==` asks whether they hold the same items. To make a real copy, use a full slice `[:]` or `list`.

````python
a = [1, 2, 3]
b = a
b.append(4)
print(a)        # a changed too
print(a is b)   # same list

c = a[:]        # a copy
c.append(99)
print(a)        # unchanged
print(a == c)   # different items now
print(a is c)   # different lists
````

## Lists as arguments

When you pass a list to a function and the function changes it, the change sticks, because the function works on the same list.

````python
def add_item(lst, item):
    lst.append(item)

shopping = ["eggs"]
add_item(shopping, "milk")
print(shopping)
````

## Tuples

A tuple is like a list, but you write it with parentheses and you cannot change it after you make it. It is useful when a group of values should stay together and fixed.

````python
point = (3, 4)
print(point)
print(point[0])
print(point[-1])
print(len(point))

print("rock" in ("rock", "paper", "scissors"))
````

A tuple is immutable, so assigning to an item is an error. The line below would raise `TypeError` if you ran it:

```python
point[0] = 10
```

## Unpacking a tuple

You can give a name to each part of a tuple in one step. This also lets you swap two values without a temporary variable.

````python
point = (3, 4)
x, y = point
print(x)
print(y)

a = 1
b = 2
a, b = b, a
print(a, b)
````

## Returning more than one value

A function can return a tuple, which is how you return several values at once.

````python
def min_and_max(numbers):
    return min(numbers), max(numbers)

low, high = min_and_max([4, 8, 15, 16, 23, 42])
print(low)
print(high)
````

## List or tuple?

Use a list when the collection will grow or change. Use a tuple for a small fixed group, like a point `(x, y)` or a pair of values you return from a function.
