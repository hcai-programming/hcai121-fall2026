### Question 15  (5 points)

The following code has several syntax errors.  Circle *all* of them.

```python
def grade(score)
    if score = 100:
        print("perfect!")
    elif score >= 60
        return "pass"
    else:
        return "fail"

print(grade(85)
print(grade(100))
```

````python
def grade(score)
    if score = 100:
        print("perfect!")
    elif score >= 60
        return "pass"
    else:
        return "fail"

print(grade(85)
print(grade(100))
````

**Answer.**

Four syntax errors:

1. `def grade(score)` is missing the colon → `def grade(score):`
2. `if score = 100:` uses `=` (assignment); a comparison needs `==` → `if score == 100:`
3. `elif score >= 60` is missing the colon → `elif score >= 60:`
4. `print(grade(85)` is missing a closing parenthesis → `print(grade(85))`

````python
def grade(score):
    if score == 100:
        print("perfect!")
    elif score >= 60:
        return "pass"
    else:
        return "fail"

print(grade(85))
print(grade(100))
````

### Question 16  (5 points)

This code runs, but there is a bug.  The expected output is `[90, 85]`, but the actual output is `[90]`.  Explain how to fix the code.


````python
def add_score(scores, s):
    scores = scores + [s]
grades = [90]
add_score(grades, 85)
print(grades)
````

**Answer.**

`scores = scores + [s]` builds a **new** list and points the local name `scores` at it; the caller's list is never touched, and the function returns `None`.  So `grades` stays `[90]`.

Fix: change the list in place, or return it:

```python
def add_score(scores, s):
    scores.append(s)
```

or

```python
def add_score(scores, s):
    return scores + [s]

grades = add_score(grades, 85)
```

````python
def add_score_correct1(scores, s):
    scores.append(s)
grades = [90]
add_score_correct1(grades, 85)
print(grades)

print("\n")
def add_score_correct2(scores, s):
    return scores + [s]
grades = [90]
grades = add_score_correct2(grades, 85)
print(grades)
````

### Question 17  (5 points)

This code runs, but there is a bug.  The function should return the running totals `[10, 30, 60]`, but it actually returns `[10, 20, 30]`.  Explain how to fix the code.


````python
def running_totals(prices):
    totals = []
    for p in prices:
        total = 0
        total = total + p
        totals.append(total)
    return totals

print(running_totals([10, 20, 30]))
````

`total = 0` is **inside** the loop, so it resets to 0 on every pass and each entry is just that one price.  Move `total = 0` above the loop so it keeps accumulating:


````python
def running_totals_correct(prices):
    totals = []
    total = 0
    for p in prices:
        total = total + p
        totals.append(total)
    return totals

print(running_totals_correct([10, 20, 30]))
````

### Question 18  (5 points)

Complete the code in the third box so that it outputs the number of player wins followed by the number of computer wins after the match keeps playing until the player or the computer reaches 10 wins.


````python
import random
def random_move():
    n = random.randint(1, 3)
    if n == 1:
        return "rock"
    elif n == 2:
        return "paper"
    else:
        return "scissors"

def winner(player, computer):
    if player == computer:
        return "tie"
    elif (player == "rock" and computer == "scissors") or \
         (player == "paper" and computer == "rock") or \
         (player == "scissors" and computer == "paper"):
        return "player"
    else:
        return "computer"
````

````python
a = 0
b = 0

while a < 10 and b < 10:

    c_move = random_move()
    p_move = random_move()
    result = winner(p_move, c_move)

    if result == "player":
        a = a + 1

    elif result == "computer":
        b = b + 1

print(a, b)
````

### Question 19  (5 points)

Consider the following code.  As in Project 1, `round_number` is the current round and `last_computer_move` is the move made in the previous round by the computer.  Explain what this function does.


````python
def strat(round_number, last_computer_move):
    if last_computer_move == "paper":
        choice = random.randint(1, 3)
        if choice == 1:
            return "rock"
        elif choice == 2:
            return "scissors"
        else:
            return "paper"
    elif last_computer_move == "scissors":
        if random.randint(1, 2) == 1:
            return "paper"
        else:
            return "rock"
    else:
        return "paper"
````

The function chooses the player's next move based only on the computer's **last move** (it ignores `round_number`):

- If the computer last played `"paper"`: it returns `"rock"`, `"scissors"`, or `"paper"`, chosen at random (each with probability 1/3).
- If the computer last played `"scissors"`: it returns `"paper"` or `"rock"`, a 50/50 choice.
- Otherwise (the computer last played `"rock"`, or there is no last move yet): it always returns `"paper"`.
