## 1. Lists of tuples

A tuple groups a few values together.  A **list of tuples** is one of the most
common ways to store records, like `(name, score)` pairs.

```python
scores = [("ana", 88), ("ben", 92), ("cara", 79)]
print(scores[0])        # the first tuple
print(scores[0][0])     # the name in the first tuple
print(scores[0][1])     # the score in the first tuple
name, score = scores[0]  # unpacking the first tuple
print(name, score)
```

**Iterate** over it by unpacking each tuple into two names:

```python
for name, score in scores:
    print(name, "scored", score)
```

**Sort** it.  By default Python sorts by the **first** element of each tuple
(here, the name):

```python
print(sorted(scores))
```

To sort by the **second** element (the score), you give `sorted` a `key`:
a function that takes one item and returns the value to sort by.

### A quick word on `lambda`

A `lambda` is a short way to write a small function with no name.
`lambda inputs: result` takes the inputs and returns the result.  The two
`sorted` calls below do exactly the same thing: the first passes a normal
function, the second writes it inline as a `lambda`.

```python
def by_score(pair):
    return pair[1]

print(sorted(scores, key=by_score))                # pass a named function
print(sorted(scores, key=lambda pair: pair[1]))    # the same thing, inline
```

Add `reverse=True` to go from high to low:

```python
print(sorted(scores, key=lambda pair: pair[1], reverse=True))
```

A `key` works on any list.  For example, sort words by length:

```python
words = ["banana", "fig", "cherry"]
print(sorted(words, key=len))
```

## 2. Dictionaries

A dictionary stores **key → value** pairs.  You look things up by key
(instead of by position like a list).

```python
ages = {"ana": 20, "ben": 19, "cara": 21}
print(ages["ana"])
print(ages["cara"])
```

Add a new key, or change an existing one:

```python
ages["dan"] = 22      # add a new key
ages["ana"] = 25      # change an existing value
print(ages)
```

Looking up a missing key is an error.  Use `.get` to avoid it:

```python
# ages["zoe"]            # this would raise a KeyError
print(ages.get("zoe"))          # None, no error
print(ages.get("zoe", 0))       # 0, a default value
```

`in` checks the **keys** of a dictionary:

```python
print("ana" in ages)
print("zoe" in ages)
```

### .items() gives you a list of (key, value) tuples

This is the bridge back to Section 1.  `.items()` hands you `(key, value)`
pairs, so you can **iterate** and **sort** them exactly like a list of tuples.

```python
for name, age in ages.items():
    print(name, "is", age)

print(list(ages.items()))
```

`.keys()` gives you just the keys, and `.values()` just the values:

```python
print(list(ages.keys()))      # just the keys
print(list(ages.values()))    # just the values
```

### The counting pattern

Count how many times each thing appears.  This shows up everywhere.

```python
moves = ["rock", "paper", "rock", "scissors", "rock", "paper"]

counts = {}
for move in moves:
    counts[move] = counts.get(move, 0) + 1
print(counts)
```

Now combine the two ideas: sort `.items()` by the value to find the most
common move (just like sorting `(name, score)` pairs by the score):

```python
ranked = sorted(counts.items(), key=lambda pair: pair[1], reverse=True)
print(ranked)
print("most common:", ranked[0][0])
```

## 3. enumerate

`enumerate` gives you the index **and** the item at the same time.

```python
colors = ["red", "green", "blue"]
for i, color in enumerate(colors):
    print(i, color)
```

You can start counting from 1:

```python
for rank, color in enumerate(colors, start=1):
    print(rank, color)
```

## 4. zip

`zip` pairs up two lists, item by item, into tuples.

```python
names = ["ana", "ben", "cara"]
scores = [88, 92, 79]
for name, score in zip(names, scores):
    print(name, score)
```

A common trick: build a dictionary from two lists:

```python
grades = dict(zip(names, scores))
print(grades)
```
