# Files and CSV

A program forgets everything when it stops running.  **Files** let data live
outside the program: you can save results, and you can load data that is far
too big to type into your code (like the 13,000 words Wordle accepts).

Run each cell, but try to **predict the output before you run it**.

## 1. Writing a text file

````python
with open("notes.txt", "w") as f:
    f.write("first line\n")
    f.write("second line\n")
print("done")
````

`open(name, "w")` opens a file for **writing**: it creates the file, or erases
it if it already exists.  `with` closes the file for you when the block ends.
`"\n"` is the newline character; without it, everything ends up on one line.

## 2. Reading a text file

````python
with open("notes.txt", "r") as f:
    text = f.read()
print(text)
````

Usually you want the file one line at a time:

````python
with open("notes.txt") as f:
    for line in f:
        print(repr(line))
````

Each line still ends with `"\n"` (that is what `repr` shows).  Use `.strip()`
to remove it:

````python
with open("notes.txt") as f:
    lines = [line.strip() for line in f]
print(lines)
````

## 3. Appending

````python
with open("notes.txt", "a") as f:
    f.write("third line\n")

with open("notes.txt") as f:
    print(f.read())
````

`"w"` starts the file over; `"a"` adds to the end.  Mixing these up is a classic
way to lose data.

## 4. CSV files

A **CSV** (comma-separated values) file is a table stored as text: one row per
line, columns separated by commas, and usually a header row with the column
names.

```
name,score
ana,88
ben,92
```

Spreadsheet programs (Excel, Google Sheets, Numbers) open CSV files directly.
Python's `csv` module reads and writes them.

### Writing a CSV file

````python
import csv

rows = [("ana", 88), ("ben", 92), ("cara", 79)]

with open("scores.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["name", "score"])     # the header row
    for name, score in rows:
        writer.writerow([name, score])
print("wrote scores.csv")
````

(`newline=""` stops the `csv` module from adding blank lines on some
computers.)  Look at what is actually in the file:

````python
with open("scores.csv") as f:
    print(f.read())
````

### Reading a CSV file

````python
with open("scores.csv") as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)
````

Each row comes back as a **list of strings**, including the header row.
Notice that `'88'` is a string, not a number.  What does this print?

````python
with open("scores.csv") as f:
    reader = csv.reader(f)
    next(reader)                  # skip the header row
    total = 0
    for name, score in reader:
        total = total + int(score)    # convert the string to an int first
print(total)
````

Everything you read from a file is text.  Use `int(...)` or `float(...)`
before doing arithmetic.

### Rows as dictionaries

`csv.DictReader` uses the header row to turn each row into a **dictionary**:

````python
with open("scores.csv") as f:
    for row in csv.DictReader(f):
        print(row["name"], "scored", row["score"])
````

A very common pattern: load a two-column CSV into a dictionary.

````python
grades = {}
with open("scores.csv") as f:
    for row in csv.DictReader(f):
        grades[row["name"]] = int(row["score"])

print(grades)
best = max(grades.items(), key=lambda pair: pair[1])
print("best:", best)
````
