# Project: Wordle Scoring

Our goal for today: write the function that **scores a Wordle guess**.  You give
it the answer and a guess, and it returns which letters are green, yellow, or
gray.

- 🟩 green: right letter, right spot
- 🟨 yellow: right letter, wrong spot
- ⬜ gray: letter not in the word

This pulls together lists of tuples, `zip` / `enumerate`, membership with `in`,
and (for the tricky version) the dictionary counting pattern.

## Setup (run first)

This `show` helper is given to you.  It turns a list of `(letter, status)`
tuples into Wordle squares.

```python
BOXES = {"green": "🟩", "yellow": "🟨", "gray": "⬜"}

def show(feedback):
    # feedback is a list of (letter, status) tuples
    print(" ".join(BOXES[status] + letter.upper() for letter, status in feedback))

# demo with a hand-made feedback list
show([("c", "green"), ("r", "yellow"), ("a", "gray"), ("n", "gray"), ("e", "green")])
```
