### Question 1

The following function has a bug.  It should output `["crane", "house", "share"]` (the list of words), but instead the output is `["crane\n", "house\n", "share\n"]` (each word followed by `\n`).  What should you change to fix the code?

````python
def load_words(word_list_file):
    with open(word_list_file) as f:
        words = [line for line in f]
    return words

word_list = load_words("words.txt")
print(word_list)
````

**Answer.**  Each line read from a file still ends with the newline character `"\n"`.  Remove it with `.strip()`:

````python
def load_words(word_list_file):
    with open(word_list_file) as f:
        words = [line.strip() for line in f]
    return words

word_list = load_words("words.txt")
print(word_list)
````

### Question 2

The following code does not run because there is an error.  Explain the error and how to fix it.

````python
import csv
# scores.csv contains: name,score / ana,88 / ben,92
with open("scores.csv") as f:
    reader = csv.reader(f)
    next(reader)
    total = 0
    for name, score in reader:
        total = total + score
print(total)
````

**Answer.**  `csv.reader` gives every value as a string, so `score` is `"88"`, not `88`.  The line `total = total + score` tries to add an `int` and a `str`, which raises a `TypeError`.  Fix: convert the score to an integer first.

````python
import csv
# scores.csv contains: name,score / ana,88 / ben,92
with open("scores.csv") as f:
    reader = csv.reader(f)
    next(reader)
    total = 0
    for name, score in reader:
        total = total + int(score)
print(total)
````

### Question 3

Before the following code runs, `notes1.txt` and `notes2.txt` each contain exactly one line: `first line`.  What does each file contain after the code runs?  Explain why the two files end up different.

First create the two files:

````python
for name in ["notes1.txt", "notes2.txt"]:
    with open(name, "w") as f:
        f.write("first line\n")
````

The quiz code:

````python
with open("notes1.txt", "a") as f:
    f.write("another line\n")

with open("notes2.txt", "w") as f:
    f.write("another line\n")
````

Look at the two files:

````python
for name in ["notes1.txt", "notes2.txt"]:
    print("---", name)
    with open(name) as f:
        print(f.read())
````

**Answer.**  `notes1.txt` contains `first line` and `another line`; `notes2.txt` contains only `another line`.  Mode `"a"` (append) adds to the end of the file and keeps what was already there.  Mode `"w"` (write) erases the file first, so the old line is lost.

### Question 4

What is the output of the following code?

````python
def score(answer, guess):
    result = []
    for g, a in zip(guess, answer):
        if g == a:
            result.append((g, "green"))
        elif g in answer:
            result.append((g, "yellow"))
        else:
            result.append((g, "gray"))
    return result

print(score("crane", "share"))
print(score("poppy", "apple"))
````

**Answer.**

```
[('s', 'gray'), ('h', 'gray'), ('a', 'green'), ('r', 'yellow'), ('e', 'green')]
[('a', 'gray'), ('p', 'yellow'), ('p', 'green'), ('l', 'gray'), ('e', 'gray')]
```

`zip(guess, answer)` pairs each letter of the guess with the letter of the answer in the same position.  Same letter: green.  Otherwise, if the letter appears anywhere in the answer: yellow.  Otherwise: gray.

- `crane` / `share`: `s` and `h` are not in `crane`; `a` and `e` match their positions; `r` is in `crane` but in a different position.
- `poppy` / `apple`: `a`, `l`, `e` are not in `poppy`; the first `p` is paired with `o`, but `p` is in `poppy`, so yellow; the second `p` is paired with `p`, so green.

### Question 5

What is the output of the following code?

````python
import csv

def save_counts(counts, filename):
    with open(filename, "w", newline="") as f:
        writer = csv.writer(f)
        writer.writerow(["move", "count"])
        for move, count in counts.items():
            writer.writerow([move, count])

save_counts({"rock": 3, "paper": 2, "scissors": 1}, "counts.csv")

with open("counts.csv") as f:
    for line in f:
        print(line.strip())
````

**Answer.**

```
move,count
rock,3
paper,2
scissors,1
```

`save_counts` writes the header row and then one row for each `(move, count)` pair from `.items()`, in the order the keys were added to the dictionary.  Reading the file back prints each line with its newline removed by `.strip()`.
