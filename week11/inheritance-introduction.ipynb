# Introduction to Subclasses and Inheritance

```python
class Shape:
    def __init__(self, name):
        self.name = name

    def sides(self):
        return 0

class Square(Shape):
    def sides(self):
        return 4

class Circle(Shape):
    pass

for s in [Square("box"), Circle("ring"), Shape("blob")]:
    print(s.name, s.sides())
```

```python
class Greeter:
    def greet(self, name):
        return "Hello, " + name

class LoudGreeter(Greeter):
    def greet(self, name):
        return super().greet(name).upper() + "!"

print(Greeter().greet("ana"))
print(LoudGreeter().greet("ana"))
```

```python
# This produces an error.
class Account:
    def __init__(self, owner):
        self.owner = owner
        self.balance = 0

class Savings(Account):
    def __init__(self, owner, rate):
        self.rate = rate

s = Savings("ana", 0.05)
print(s.owner, s.rate)
```

`Savings` has its own `__init__`, which replaces `Account.__init__`.  So the lines `self.owner = owner` and `self.balance = 0` never run, and `s.owner` raises an `AttributeError`.  The fix is to call the parent's `__init__` with `super()`:

```python
class Account:
    def __init__(self, owner):
        self.owner = owner
        self.balance = 0

class Savings(Account):
    def __init__(self, owner, rate):
        super().__init__(owner)
        self.rate = rate

s = Savings("ana", 0.05)
print(s.owner, s.rate, s.balance)
```

```python
class Player:
    def __init__(self, name):
        self.name = name

    def next_guess(self, history):
        return "crane"
```

Write a subclass `FixedPlayer` of `Player`.  Its `__init__` method takes a `name` and a `word`, uses `super()` to store the name, and stores the word as an attribute.  It overrides `next_guess(history)` so that it always returns the stored word.  Then write one line that creates a `FixedPlayer` named `"stubborn"` whose word is `"slate"`.

```python
class FixedPlayer(Player):
    def __init__(self, name, word):
        super().__init__(name)
        self.word = word

    def next_guess(self, history):
        return self.word

p = FixedPlayer("stubborn", "slate")
print(p.name, p.next_guess([]))
```
