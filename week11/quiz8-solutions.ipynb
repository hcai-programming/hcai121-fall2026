### Question 1

Consider the following definition of a `Playlist` class:

```python
class Playlist:
    def __init__(self, name):
        self.name = name
        self.songs = []

    def add(self, song):
        self.songs.append(song)
```

**(a)** Using the `Playlist` class, what is the output of the following code?

```python
a = Playlist("road trip")
b = Playlist("study")
a.name = "summer"
print(a.name, a.songs)
print(b.name, b.songs)
```

`a` and `b` are two separate objects.  Each one gets its own `name` and its own `songs` list when `__init__` runs.  Changing `a.name` does not change `b.name`, and no songs have been added yet.

**(b)** Using the `Playlist` class, what is the output of the following code?

```python
a = Playlist("road trip")
b = Playlist("study")
a.add("song 1")
b.add("song 2")
a.add("song 3")
print(a.name, a.songs)
print(b.name, b.songs)
```

`a.add("song 1")` runs `add` with `self` equal to `a`, so the song goes into `a.songs`.  Each playlist has its own list.

### Question 2

The following code does not run because there is an error.  Explain the error and how to fix it.

```python
class Player:
    def __init__(self, name):
        self.name = name
        wins = 0

    def record_win(self):
        self.wins = self.wins + 1

p = Player("ana")
p.record_win()
print(p.name, p.wins)
```

**Answer.**  In `__init__`, `wins = 0` creates a local variable that disappears when `__init__` ends.  It is not stored in the object, so `self.wins` does not exist when `record_win` tries to read it, and Python raises an `AttributeError`.  Fix: write `self.wins = 0`.

```python
class Player:
    def __init__(self, name):
        self.name = name
        self.wins = 0

    def record_win(self):
        self.wins = self.wins + 1

p = Player("ana")
p.record_win()
print(p.name, p.wins)
```

### Question 3

The following code does not run because there is an error.  Explain the error and how to fix it.

```python
import random

class RandomPlayer:
    def __init__(self, name, words):
        self.name = name
        self.words = words

    def next_guess(history):
        return random.choice(self.words)

p = RandomPlayer("random", ["crane", "slate", "ghost"])
print(p.next_guess([]))
```

**Answer.**  Every method needs `self` as its first parameter.  `p.next_guess([])` passes two things, the object `p` and the list `[]`, but `next_guess(history)` only accepts one, so Python raises a `TypeError`.  Fix: `def next_guess(self, history):`.

```python
import random

class RandomPlayer:
    def __init__(self, name, words):
        self.name = name
        self.words = words

    def next_guess(self, history):
        return random.choice(self.words)

p = RandomPlayer("random", ["crane", "slate", "ghost"])
print(p.next_guess([]))
```

### Question 4

Write a class `FixedPlayer`.  Its `__init__` method takes a `name` and a `word` and stores both as attributes.  It has a method `next_guess(history)` that always returns the stored word.  Then write one line that creates a `FixedPlayer` named `"stubborn"` whose word is `"crane"`.

**Answer.**

```python
class FixedPlayer:
    def __init__(self, name, word):
        self.name = name
        self.word = word

    def next_guess(self, history):
        return self.word

p = FixedPlayer("stubborn", "crane")
print(p.name, p.next_guess([]))
```
